```python
filepath = os.path.join("..", "..", "Zhaopings_V1_code", "purecodes")

N_x, N_y, NC = 100, 60, 100
K = 12

input_flat = np.loadtxt(os.path.join(filepath, "InputIr.csv"), delimiter=",")
input_array = input_flat.reshape((N_x, N_y, K), order='C')
input_array = np.swapaxes(input_array, 0, 1)

angles = np.linspace(0, np.pi, K, endpoint=False)
M = angles[np.newaxis, np.newaxis, :]  # shape (1, 1, K)
A = np.broadcast_to(M, (N_y, N_x, K))
visualize_input(A, input_array, dpi=100)
plt.show()
```

```python
output_flat = np.loadtxt(os.path.join(filepath, "outSp_over_time_T=30_I=4_dt=0.01.csv"), delimiter=",")
```

```python
totaltime = 30.0
dt = 0.01
T = int(totaltime / dt) 
model_output = output_flat.reshape((T, N_x, N_y, K), order='C')
model_output = np.swapaxes(model_output, 1, 2)
```

```python
print(np.all(model_output == 0.))
print(np.min(np.nonzero(input_array[31, :])[0]))  # first non-zero index

C_out_half = model_output.mean(axis=0)
A_out_half = np.broadcast_to(M, (N_y, N_x, K))
visualize_output(A_out_half[25:-25, 40:-40], C_out_half[25:-25, 40:-40] * 1, verbose=False, dpi = 500)
plt.show()
```
